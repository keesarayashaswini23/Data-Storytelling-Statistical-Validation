```python
# ============================================
# TASK 4 - Hypothesis Testing
# Data Analytics Internship
# ============================================

# Step 1: Import Required Libraries

import pandas as pd
from scipy.stats import ttest_ind

# ============================================
# Step 2: Load the Cleaned Dataset
# ============================================

df = pd.read_excel("Cleaned_Sales_Dataset.xlsx")

# Display first 5 rows
print("First 5 Rows of Dataset:")
print(df.head())

# ============================================
# Step 3: Business Question
# ============================================

print("\nBusiness Question:")
print("Is there a significant difference in Total Sales between male and female customers?")

# ============================================
# Step 4: Define Hypotheses
# ============================================

print("\nNull Hypothesis (H₀):")
print("There is no significant difference in Total Sales between male and female customers.")

print("\nAlternative Hypothesis (H₁):")
print("There is a significant difference in Total Sales between male and female customers.")

# ============================================
# Step 5: Separate the Data
# ============================================

male_sales = df[df["Gender"] == "Male"]["Total_Sales"]

female_sales = df[df["Gender"] == "Female"]["Total_Sales"]

# ============================================
# Step 6: Perform Independent Sample t-test
# ============================================

t_stat, p_value = ttest_ind(male_sales, female_sales)

# ============================================
# Step 7: Display Results
# ============================================

print("\n===== Hypothesis Test Result =====")

print("T-Statistic :", t_stat)
print("P-Value :", p_value)

# ============================================
# Step 8: Conclusion
# ============================================

if p_value < 0.05:
    print("\nDecision:")
    print("Reject the Null Hypothesis")

    print("\nConclusion:")
    print("There is a statistically significant difference in Total Sales between male and female customers.")

else:
    print("\nDecision:")
    print("Fail to Reject the Null Hypothesis")

    print("\nConclusion:")
    print("There is no statistically significant difference in Total Sales between male and female customers.")
```
